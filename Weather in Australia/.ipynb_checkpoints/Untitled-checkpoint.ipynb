```python
# finding null% in whole dataframe

print("% of data is null")
(df.isna().sum() / df.shape[0]*100).sort_values(ascending=False)
```

```python
# keeping usefull columns only (deleting date column)
df = df.drop(columns="date")
```

```python
# removing those rows where target is null ie. rain today and rain tommrow
df = df[(df['raintoday'].notna() ) & (df['raintomorrow'].notna())]
```

```python
df.isna().sum()
```

```python
# finding columns names for category and numericals

column_names = numerical_and_categorical_columns(df)
```

```python
# filling null values in numerical columns

for i in numeric_columns:
    df[i] =df[i].fillna(df[i].mean())
```

```python
# checking there are null values in categorical columns

df[column_names['categorical_columns']].isnull().sum()
```

```python
# filling those null values in categorical columns
df[column_names['categorical_columns']] = df[column_names['categorical_columns']].fillna("unknown")
```

```python
df.info()
```

```python
# finding all numerical and categorical columns
def numerical_and_categorical_columns(df):
    list_of_columns = df.columns
    numeric_columns = []
    categorical_columns = []
    for i in list_of_columns:
        if df[i].dtype == int or df[i].dtype == float:
            numeric_columns.append(i)
        else:
            categorical_columns.append(i)
    return {"numerical_columns":numeric_columns,"categorical_columns":categorical_columns}
        
```
