```python
df['gender'] = df['origin'].map({1:"male",2:"femail",3:"male"})
```

```python
df['yesno'] = df['cylinders'] >5
```

```python
df['cylinders'] =  df['cylinders'].astype(str)
```

```python
d = pd.crosstab(df['gender'],df['yesno'])
```

```python
rt = d.sum(axis=1)
ct = d.sum(axis=0)
```

```python
rt
```

```python
ct
```

```python
alltotal = sum(rt)+sum(ct)
```

```python
alltotal
```

```python
expected = np.outer(rt,ct) /alltotal
```

```python
expected
```

```python
observed = np.array(d)
```

```python
chi_square = np.array((observed - expected) ** 2 / expected).sum()
print("Chi-Square Statistic:", chi_square)
```
