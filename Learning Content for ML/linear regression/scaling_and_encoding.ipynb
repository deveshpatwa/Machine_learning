## Data Understanding

| Column        | Description                                          |
| ------------- | ---------------------------------------------------- |
| `survived`    | Survival status (0 = No, 1 = Yes)                    |
| `pclass`      | Passenger class (1 = First, 2 = Second, 3 = Third)   |
| `sex`         | Gender of the passenger                              |
| `age`         | Age in years                                         |
| `sibsp`       | Number of siblings/spouses aboard                    |
| `parch`       | Number of parents/children aboard                    |
| `fare`        | Ticket fare paid                                     |
| `embarked`    | Port of embarkation (`C`, `Q`, `S`)                  |
| `class`       | Passenger class as text (`First`, `Second`, `Third`) |
| `who`         | Passenger category (`man`, `woman`, `child`)         |
| `adult_male`  | Whether passenger is an adult male (`True`/`False`)  |
| `deck`        | Cabin deck (`A`–`G`)                                 |
| `embark_town` | Full name of embarkation town                        |
| `alive`       | Survival status as text (`yes`, `no`)                |
| `alone`       | Whether passenger traveled alone (`True`/`False`)    |
