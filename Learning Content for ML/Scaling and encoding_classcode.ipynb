```python
def is_even(n: int) -> bool:
    return n % 2 == 0

def check_user_number() -> None:
    s = input("Enter an integer: ")
    try:
        n = int(s)
    except ValueError:
        print("Please enter a valid integer.")
        return
    if is_even(n):
        print(f"{n} is even")
    else:
        print(f"{n} is odd")

if __name__ == "__main__":
    
    check_user_number()
```
